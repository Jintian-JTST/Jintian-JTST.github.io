## Plotting data

### Fitting using SciPy

Now we would like to fit our data. To do so we need to import a statistics package, of which there are many choices, we will use a common one here.
A module which makes this easy is the 'scipy.optimize' module and the 'curve_fit' function.

```python
from scipy.optimize import curve_fit
```

The 'curve_fit' function requires us to define the type of fit. 
We first define a linear fit here.

```python
def linear(x, a, b):
    return a * x + b
```

We then perform the fit with the (linear) function we have defined.
- In the 'curve_fit' function we specify our 'x' and 'y' data and also the uncertainties on our y-data (we called this variable 'yerr')
- We specify the function to fit, which we just named 'linear'.
- We must specify the 'absolute_sigma=True' option. This option changes the way the fitting function works based on the type of uncertainties we have given it. When we have uncertainties based on measurements which are the same units as our data (which will be the case for most of your lab work) we specify this option to be True.

The 'curve_fit' function returns two things:
1. The fitted set of parameters (here this contains our fitted 'a' and 'b' values).
2. The covariance matrix. (This is a matrix and gives information about how well the parameters are affected by each other.  Discussed in 2nd year stats!)

```python
parameters, covariant_matrix = curve_fit(linear, x, y, sigma=yerr, absolute_sigma=True)
slope = parameters[0] #extract slope, it is parameter #0
intercept = parameters[1] #extract intercept, it is parameter #1
```

To find the errors on the parameters, we have to take the square root of the diagonal elements of the variance  (importing the numpy library for the square root).

```python
import numpy as np
parameter_uncertainties = np.sqrt(covariant_matrix.diagonal())
slope_error = parameter_uncertainties[0]
intercept_error = parameter_uncertainties[1]

print('Intercept: {:.0f} Error on intercept: {:.0f}'.format(intercept, intercept_error))
print('Slope: {:.2f} Error on slope: {:.2f}'.format(slope, slope_error))
```

## Combining plots

Now we:
- plot both the data and the fitted line and,
- save our plot as a file using the 'savefig' function (note: must be executed from this cell)

The saved image file, 'hubble_fit.png' should be saved in the same folder as where this Python script is.

```python
fig = plt.figure(figsize=(5,4), dpi=200)
plt.errorbar(x, y, yerr, label='Data', fmt='.', capsize=5)
plt.plot(x, linear(x, slope, intercept), 'r', label='Fitted line') # note: we can write any formula into this plotting function
plt.title('Speed vs Distance')
plt.xlabel('Distance (M-Parsecs)')
plt.ylabel('Speed (km/sec)')
plt.legend()

plt.savefig('hubble_fit.png', dpi=200)
plt.show()
```

## Data re-visited

We could equally swap the data in the plot around and plot distance vs speed.
To do this we:
- re-set our x, y and yerr variables to the appriate data.
- re-calculate the fit and extract the slope and intercept

```python
x = df["Speed (km/sec)"].values
y = df["Distance (M-Parsecs)"].values
yerr = df["error in D (M-Parsecs)"].values

parameters, covariant_matrix = curve_fit(linear, x, y, sigma=yerr, absolute_sigma=True)
slope = parameters[0] #slope is parameter #0
intercept = parameters[1] #intercept is parameter #1

parameter_uncertainties = np.sqrt(covariant_matrix.diagonal())
slope_error = parameter_uncertainties[0]
intercept_error = parameter_uncertainties[1]

print('Intercept: {:.2f} Error on intercept: {:.2f}'.format(intercept, intercept_error))
print('Slope: {:.5f} Error on slope: {:.5f}'.format(slope, slope_error))
```

Since we swapped the axes, we invert the slope to compare with the previous value. 
Note, for the uncertainty, we take the fractional uncertainty and multiply by the new value (i.e. the inverted value)

```python
print('1/Slope: {:.1f} Error on slope: {:.1f}'.format(1/slope, (slope_error/slope)*1/slope))
```
