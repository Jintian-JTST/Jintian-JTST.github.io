# Some definitions


<div class="alert alert-block alert-info" style="text-align:center;">
    <b>Accuracy:</b>  The difference between the result and the true value; an accurate measurement is in agreement with the "accepted" value
</div>


<div class="alert alert-block alert-info" style="text-align:center;">
    <b>Precision:</b>  The  least distinguishable change, given by the number of digits or the spread in the measurements (precise = small spread)
</div>

# Types of error (1)

<div class="alert alert-block alert-info" style="text-align:center;">
    <b>Random errors:</b>  error associated with limitations of the measuring device, they prevent observer from obtaining same value each time they measure.  Measurements usually look like a peak centered around a value.  Example: see previous graph. 
    $\implies$ Reduce by taking more measurements.
</div>

<div class="alert alert-block alert-info" style="text-align:center;">
    <b>Mistakes:</b> e.g. wrote the number 3.75 as 3.57, typed number in wrong to computer, misread scale.  Plotting data can show up points which “don’t fit” and these can then be discarded or perhaps measured again.  Example: the point at 32.3mm
</div>

# Types of error (2)


<div class="alert alert-block alert-info" style="text-align:center;">
    <b>Systematic errors:</b>  error which always has the same sign so shifts the centre of the peak one way or another but not both.  Example: ruler calibrated at one temperature but we are using it at a higher temperature and ruler has expanded. $\implies$ Difficult to detect, have to try and design experiment to avoid them.
</div>


<div class="alert alert-block alert-info" style="text-align:center;">
    <b>Tolerances:</b> Engineering tolerances on mechanical and electrical components. Don't confuse with random error. For discussion, see aforementioned <a href="https://www-teaching.physics.ox.ac.uk/practical_course/scripts/srv/local/rscripts/trunk/Admin/AD02/AD02.pdf">AD02</a> and <a href="https://www-teaching.physics.ox.ac.uk/practical_course/scripts/srv/local/rscripts/trunk/Admin/AD26/AD26.pdf">AD26</a> scripts.
</div>

# Which types of errors dominate?

- The error which has the largest influence on your result is called the **dominant error**
- Part of your job as a physicist (especially when carrying out experiments) is to determine which is the dominant error 
    - $\implies$ reduce it as much as possible!
    - Caveat: what happens when this error becomes subdominant?
- Important when propagating errors in formulae -- if one value has an error much bigger than all the rest or appears as a power in a formula it may dominate any error calculation.

**More about propagating errors later!**

# Fundamental limits

Some uncertainties can never be eliminated:

<div class="alert alert-block alert-info" style="text-align:center;">
    <b>Shot noise:</b>  discrete nature of certain quantities, e.g. measuring a feeble light beam start to get random fluctuations of intensity because light is made up of photons (finite energy bundles)
</div>


<div class="alert alert-block alert-info" style="text-align:center;">
    <b>Johnson noise:</b> arises in a resistor as a consequence of the thermal agitation of the charge carriers.  Becomes important in experiments measuring low voltages -- so might cool down experimental setup to reduce this but cannot eliminate.
</div>

$\implies$ The best you will ever be able to do is get down to these limits, but you are often way above this.

# Going back to our piece of string...

<img src="img/string-hist.png" alt="Histogramm of string lengths" style="width: 20%; margin-left: 5%; margin-top: 3%; float:right;" />

So what answer would we quote?

Values were: 

| 31.7mm | 31.8mm <br />(x 3) | 31.9mm <br />(x 4) | 32.0mm <br />(x 2) | 32.1mm | 32.3mm <br />(discard this point) |
|:-:|:-:|:-:|:-:|:-:|:-:|

<div class="alert alert-block alert-success" style="">
    <b>Mean:</b>  for a set of $N$ measurements, each with value $x_i$, the mean $\bar{x}$ is given by:    
$$\bar{x}=\cfrac{x_1+x_2+...+x_N}{N}$$
    $\implies$ This is the result!
</div>

$\implies$ so $\bar{x}=\cfrac{31.7 + 31.8 \cdot 3 + 31.9 \cdot 4 + 32.0 \cdot 2 + 32.1}{11} ~\mathrm{mm}= 31.89090909' ~\mathrm{mm}$

- How many significant figures should we quote?  Depends on the error in this value.

# How do we find the error in the mean? (1)

We will need the **standard error** which we derive from the standard deviation:

<div class="alert alert-block alert-success">
    <b>Sample standard deviation:</b>
$$ \sigma_{N-1} = \sqrt{\cfrac{\sum(x_i-\bar{x})^2}{N-1}}$$
    
- We are squaring the differences between each point and the mean -- if we didn't square it some of the terms would be positive, some negative, and they cancel out
- We use $N-1$ as we use $N$ to calculate the mean, but are now using the mean in this formula (we have reduced the number of independent values or degrees of freedom)
</div>

# How do we find the error in the mean? (2)

$$ \sigma_{N-1} = \sqrt{\cfrac{\sum(x_i-\bar{x})^2}{N-1}}$$
- Definitions of standard deviation (sample, population, ...) can vary! --  see lecture 2 
- The square of the standard deviation ($\sigma^2$) is called the **variance**.

<div class="alert alert-block alert-success" style="">
    <b>Standard error:</b> the standard deviation of the mean. 
$$ \alpha=\cfrac{\sigma_{N-1}}{\sqrt{N}} = \sqrt{\cfrac{\sum(x_i-\bar{x})^2}{N(N-1)}}$$
    $\implies$ This represents the error in the result (mean) due to random fluctuations!
</div>

<div class="alert alert-block alert-warning" style="text-align:center;">
$\implies$ As $N$ increases, $\alpha$ decreases! So take more readings to reduce your standard error $\alpha$!
<br /><br />
    NB: the estimated standard deviation $\sigma_{N-1}$ will typically <b>not</b> reduce with increasing $N$!
</div>

# Going back to our piece of string...

$$ \begin{align*}\alpha&= \sqrt{\cfrac{\sum(x_i-\bar{x})^2}{N(N-1)}} = \sqrt{\cfrac{(31.7 - 31.89090909')^2 + ... + (32.1-31.89090909')^2}{11\cdot (11-1)}}~\mathrm{mm} \\
\implies\alpha&=0.03425717~\mathrm{mm}\end{align*}$$

So what do we quote as the result for the length of the string?

$$\implies \bar{x}\pm\alpha = (31.89090909\pm 0.03425717)~\mathrm{mm}$$

What about the number of significant figures?  Only quote as many in the mean as is sensible given the error (typically quoted to **1 significant figure**):

$$\underline{\underline{(31.89\pm 0.03)~\mathrm{mm}}}\qquad \text{or}\qquad \underline{\underline{31.89(3)~\mathrm{mm}}}$$ 

<div class="alert alert-block alert-warning" style="text-align:center;">
    <b>Don't forget the units!</b>
</div>

# A slight caveat

- Sometimes you don't get such a spread of numbers, e.g. when always getting the same value while measuring with some digital calipers or a digital multimeter

- The error in the value is then not given by the formulae on previous slides but related to the resolution of the measuring device (the **tolerance!**)
    - $\implies$ Quote this as the error instead!

# Standard error visualisation (4)

<div class="alert alert-block alert-warning" style="text-align:center;">
    $\implies$ The distribution of means becomes narrower with increasing sample size $N$ as expected (increased precision!):
    $$ \alpha = \cfrac{\sigma_{N-1}}{\sqrt{N}}$$
